```python
# Output Guide:
# - Prints input/context shapes, attention row sums, and one attention row.
# - Correctness check: each row sum should be ~1.0 (softmax normalization).
# - Attention weights indicate how strongly each token attends to the others.
"""
Transformer — Scaled Dot-Product Self-Attention From Scratch
=============================================================
Implements single-head self-attention:

    Q = X @ W_q          (queries)
    K = X @ W_k          (keys)
    V = X @ W_v          (values)

    Attention(Q, K, V) = softmax( Q @ K^T / sqrt(d_k) ) @ V

The division by sqrt(d_k) prevents dot products from growing large,
keeping the softmax in a well-behaved gradient region.
"""

import numpy as np

rng = np.random.default_rng(42)

SEQ_LEN = 5    # number of tokens
D_MODEL = 8    # embedding dimension


def init_projections(d_model: int = D_MODEL):
    """Initialise query, key, value projection matrices."""
    scale = 0.2
    W_q = rng.normal(scale=scale, size=(d_model, d_model))
    W_k = rng.normal(scale=scale, size=(d_model, d_model))
    W_v = rng.normal(scale=scale, size=(d_model, d_model))
    return W_q, W_k, W_v


def softmax_2d(scores: np.ndarray) -> np.ndarray:
    """Row-wise numerically stable softmax."""
    scores_shifted = scores - scores.max(axis=1, keepdims=True)
    exp_scores = np.exp(scores_shifted)
    return exp_scores / exp_scores.sum(axis=1, keepdims=True)


def scaled_dot_product_attention(X, W_q, W_k, W_v):
    """
    Compute single-head self-attention.

    Parameters
    ----------
    X   : token embeddings, shape (T, d_model)

    Returns
    -------
    context        : attended output,     shape (T, d_model)
    attention_weights : attention matrix, shape (T, T)
    """
    d_k = X.shape[1]   # key dimension for scaling

    # Project inputs into query, key, value spaces
    Q = X @ W_q        # (T, d_model)
    K = X @ W_k        # (T, d_model)
    V = X @ W_v        # (T, d_model)

    # Compute raw attention scores and scale
    scores = Q @ K.T / np.sqrt(d_k)   # (T, T)

    # Softmax → attention weights (each row sums to 1)
    attention_weights = softmax_2d(scores)   # (T, T)

    # Weighted sum of values
    context = attention_weights @ V          # (T, d_model)

    return context, attention_weights


if __name__ == "__main__":
    X = rng.normal(size=(SEQ_LEN, D_MODEL))    # synthetic token embeddings
    W_q, W_k, W_v = init_projections()

    context, attn = scaled_dot_product_attention(X, W_q, W_k, W_v)

    print(f"Input shape     : {X.shape}  (tokens x d_model)")
    print(f"Context shape   : {context.shape}")
    print(f"Attention row sums (should all be ~1.0):")
    print(f"  {np.round(attn.sum(axis=1), 4)}")
    print(f"Attention weights (token 0 attending to all tokens):")
    print(f"  {np.round(attn[0], 4)}")

```
