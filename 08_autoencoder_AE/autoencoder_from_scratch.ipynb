```python
# Output Guide:
# - Prints training start, reconstruction MSE, and one latent code vector.
# - Correctness check: reconstruction MSE should decrease to a reasonably small value.
# - Lower MSE means better input reconstruction; latent code is the compressed representation.
"""
Autoencoder (AE) — NumPy From Scratch
======================================
Architecture : Input(6) → Encoder → Bottleneck(2) → Decoder → Output(6)
Activation   : Sigmoid (encoder and decoder)
Loss         : Mean Squared Error (reconstruction)
Training     : Backpropagation via gradient descent

The network learns a compressed 2-D representation of 6-D inputs.
"""

import numpy as np

rng = np.random.default_rng(42)

INPUT_DIM  = 6
LATENT_DIM = 2   # bottleneck size
EPOCHS     = 1000
LR         = 0.5


def sigmoid(z: np.ndarray) -> np.ndarray:
    """Element-wise sigmoid activation."""
    return 1.0 / (1.0 + np.exp(-z))


def sigmoid_grad(a: np.ndarray) -> np.ndarray:
    """Derivative of sigmoid given its OUTPUT a: a * (1 - a)."""
    return a * (1.0 - a)


def init_weights():
    scale = 0.3
    W1 = rng.normal(scale=scale, size=(INPUT_DIM,  LATENT_DIM))   # encoder weights
    b1 = np.zeros(LATENT_DIM)
    W2 = rng.normal(scale=scale, size=(LATENT_DIM, INPUT_DIM))    # decoder weights
    b2 = np.zeros(INPUT_DIM)
    return W1, b1, W2, b2


def forward(X, W1, b1, W2, b2):
    """
    Encode then decode.

    Returns
    -------
    code : latent representation  (N, LATENT_DIM)
    recon: reconstructed output   (N, INPUT_DIM)
    """
    code  = sigmoid(X @ W1 + b1)    # encoder: compress
    recon = sigmoid(code @ W2 + b2) # decoder: reconstruct
    return code, recon


def train(X, W1, b1, W2, b2, epochs=EPOCHS, lr=LR):
    """
    Backpropagate MSE reconstruction loss through both layers.

    Gradients:
        dL/d(recon) = (recon - X) / N        ← MSE gradient
        × sigmoid'(recon)                     ← through decoder sigmoid
    Then standard chain rule back through W2 → sigmoid → W1.
    """
    n = len(X)

    for epoch in range(epochs):
        # ── Forward ──────────────────────────────────────────────────────
        code, recon = forward(X, W1, b1, W2, b2)

        # ── Output gradient (MSE + decoder sigmoid) ───────────────────────
        d_out = (recon - X) * sigmoid_grad(recon) / n  # (N, INPUT_DIM)

        # ── Decoder weight gradients ──────────────────────────────────────
        dW2 = code.T @ d_out                           # (LATENT_DIM, INPUT_DIM)
        db2 = d_out.sum(axis=0)                        # (INPUT_DIM,)

        # ── Backprop through encoder sigmoid ──────────────────────────────
        d_code = (d_out @ W2.T) * sigmoid_grad(code)  # (N, LATENT_DIM)

        # ── Encoder weight gradients ──────────────────────────────────────
        dW1 = X.T @ d_code                             # (INPUT_DIM, LATENT_DIM)
        db1 = d_code.sum(axis=0)                       # (LATENT_DIM,)

        # ── Gradient descent update ───────────────────────────────────────
        W2 -= lr * dW2;  b2 -= lr * db2
        W1 -= lr * dW1;  b1 -= lr * db1

    return W1, b1, W2, b2


if __name__ == "__main__":
    X = rng.random((120, INPUT_DIM))   # random data in [0, 1]
    W1, b1, W2, b2 = init_weights()

    print(f"Training autoencoder ({INPUT_DIM} -> {LATENT_DIM} -> {INPUT_DIM}) ...")
    W1, b1, W2, b2 = train(X, W1, b1, W2, b2)

    code, recon = forward(X, W1, b1, W2, b2)
    mse = float(np.mean((recon - X) ** 2))

    print(f"Reconstruction MSE : {mse:.5f}")
    print(f"Latent code (first sample): {np.round(code[0], 3)}")

```
