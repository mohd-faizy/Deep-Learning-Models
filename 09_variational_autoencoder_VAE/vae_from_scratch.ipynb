```python
# Output Guide:
# - Prints latent samples z, encoder means mu, KL per sample, and mean KL.
# - Correctness check: KL values should be non-negative (or numerically very close).
# - KL quantifies how far latent posterior is from a standard normal prior.
"""
Variational Autoencoder (VAE) — NumPy From Scratch
====================================================
The VAE differs from a standard autoencoder in that the encoder outputs
a distribution (mean μ and log-variance log σ²) rather than a single point.

Reparameterisation trick:
    z = μ + σ * ε,   where ε ~ N(0, I)

This makes sampling differentiable so gradients can flow back to μ and log σ².

KL divergence term (regularises the latent space toward N(0, 1)):
    KL(q(z|x) || p(z)) = -0.5 * Σ [ 1 + log σ² - μ² - σ² ]
"""

import numpy as np

rng = np.random.default_rng(42)

BATCH_SIZE = 4
LATENT_DIM = 2


def encode(x_batch):
    """
    Dummy encoder: returns μ and log σ² for each sample.
    In a full VAE these would come from learned neural network layers.
    """
    mu     = rng.normal(size=(BATCH_SIZE, LATENT_DIM))
    logvar = rng.normal(scale=0.2, size=(BATCH_SIZE, LATENT_DIM))
    return mu, logvar


def reparameterise(mu, logvar):
    """
    Sample latent vectors using the reparameterisation trick.

    z = mu + std * epsilon,  std = exp(0.5 * logvar)
    """
    std     = np.exp(0.5 * logvar)          # convert log-variance to std dev
    epsilon = rng.normal(size=mu.shape)     # random noise
    z       = mu + std * epsilon            # differentiable sample
    return z


def kl_divergence(mu, logvar):
    """
    KL divergence from N(mu, sigma^2) to N(0, 1), per sample.

    KL = -0.5 * sum_j [ 1 + logvar_j - mu_j^2 - exp(logvar_j) ]
    """
    kl_per_dim = 1.0 + logvar - mu ** 2 - np.exp(logvar)
    kl_per_sample = -0.5 * kl_per_dim.sum(axis=1)
    return kl_per_sample


if __name__ == "__main__":
    # Encode a batch into distribution parameters
    mu, logvar = encode(None)

    # Sample latent vectors via reparameterisation
    z = reparameterise(mu, logvar)

    # Compute KL divergence for each sample
    kl = kl_divergence(mu, logvar)

    print(f"Latent samples z  :\n{np.round(z, 3)}")
    print(f"\nmu (encoder mean) :\n{np.round(mu, 3)}")
    print(f"\nKL divergence per sample: {np.round(kl, 3)}")
    print(f"Mean KL (regularisation loss): {kl.mean():.4f}")

```
