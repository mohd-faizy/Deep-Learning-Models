```python
# Output Guide:
# - Prints target distribution stats and generated distribution stats.
# - Correctness check: generated mean/std should move toward target mean/std after training.
# - Closer generated stats signify the generator is matching the data distribution better.
"""
Generative Adversarial Network (GAN) — NumPy From Scratch
==========================================================
A minimal 1-D GAN that learns to match a Gaussian target N(2, 0.5).

Components:
  Generator  G(z) = g_w * z + g_b   (linear: maps noise → fake sample)
  Discriminator D(x) = sigmoid(d_w * x + d_b)  (binary classifier: real vs fake)

Objective (non-saturating):
  D maximises: E[log D(real)] + E[log(1 - D(fake))]
  G minimises: E[log(1 - D(G(z)))]  ≡ maximises E[log D(G(z))]
"""

import numpy as np

rng = np.random.default_rng(42)

BATCH_SIZE = 64
EPOCHS     = 800
LR         = 0.03

# Target distribution: real samples come from N(2, 0.5)
TARGET_MEAN = 2.0
TARGET_STD  = 0.5


def sigmoid(x: np.ndarray) -> np.ndarray:
    return 1.0 / (1.0 + np.exp(-x))


def train_gan():
    # ── Initialise parameters ─────────────────────────────────────────────
    g_w, g_b = 0.5, 0.0   # generator:      fake = g_w * z + g_b
    d_w, d_b = 0.2, 0.0   # discriminator:  score = sigmoid(d_w * x + d_b)

    for epoch in range(EPOCHS):
        # ── Sample real and fake data ─────────────────────────────────────
        real = rng.normal(TARGET_MEAN, TARGET_STD, BATCH_SIZE)
        z    = rng.normal(size=BATCH_SIZE)
        fake = g_w * z + g_b

        # ── Discriminator scores ──────────────────────────────────────────
        d_real = sigmoid(d_w * real + d_b)   # want these close to 1
        d_fake = sigmoid(d_w * fake + d_b)   # want these close to 0

        # ── Discriminator gradients ───────────────────────────────────────
        # dL_D/d_w = -mean[ (1 - d_real) * real + d_fake * fake ]
        grad_d_w = -np.mean((1 - d_real) * real + d_fake * fake)
        grad_d_b = -np.mean((1 - d_real) + d_fake)

        d_w -= LR * grad_d_w
        d_b -= LR * grad_d_b

        # ── Generator gradients ───────────────────────────────────────────
        # G wants D(fake) → 1, so G minimises -log D(G(z))
        d_fake_updated = sigmoid(d_w * (g_w * z + g_b) + d_b)
        grad_g = (d_fake_updated - 1.0) * d_w    # chain rule

        g_w -= LR * np.mean(grad_g * z)           # dL_G/dg_w
        g_b -= LR * np.mean(grad_g)               # dL_G/dg_b

    return g_w, g_b


if __name__ == "__main__":
    print("Training minimal 1-D GAN …")
    g_w, g_b = train_gan()

    # Draw 1000 samples from the trained generator
    z_test = rng.normal(size=1_000)
    generated = g_w * z_test + g_b

    print(f"Target distribution : mean={TARGET_MEAN:.2f}, std={TARGET_STD:.2f}")
    print(f"Generated mean      : {generated.mean():.3f}")
    print(f"Generated std       : {generated.std():.3f}")

```
