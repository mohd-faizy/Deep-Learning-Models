```python
# Output Guide:
# - Prints input/Phi shapes, training MSE, first predictions, and first true values.
# - Correctness check: low MSE and predictions close to true values on training points.
# - Phi is the RBF feature matrix (distance-to-centers transformed features).
"""
Radial Basis Function Network (RBFN) — NumPy From Scratch
==========================================================
An RBFN replaces hidden-layer neurons with Radial Basis Functions.
Each neuron responds most strongly to inputs near its centre.

Architecture:
    Input x → RBF layer φ(x) → Linear output layer w

RBF activation:
    φ_i(x) = exp( -γ * ||x - c_i||² )

The output layer is solved analytically via the pseudoinverse
(no iterative gradient descent needed for regression).

Task: Learn to approximate f(x) = sin(2x) + 0.2x on [-3, 3].
"""

import numpy as np

# ── Data ─────────────────────────────────────────────────────────────────────
N_SAMPLES  = 80
N_CENTERS  = 12
GAMMA      = 1.2    # width parameter for RBF kernels

x      = np.linspace(-3, 3, N_SAMPLES)       # 1-D input
X      = x[:, None]                           # shape: (N, 1)
y      = np.sin(2 * x) + 0.2 * x             # regression target


def make_centers(n_centers: int = N_CENTERS) -> np.ndarray:
    """Evenly space RBF centres across the input range."""
    centers = np.linspace(-3, 3, n_centers)
    return centers[:, None]   # shape: (K, 1)


def rbf_layer(X: np.ndarray, centers: np.ndarray, gamma: float = GAMMA) -> np.ndarray:
    """
    Compute the RBF design matrix Φ.

    Φ[i, j] = exp( -gamma * ||x_i - c_j||² )

    Parameters
    ----------
    X       : input data,   shape (N, 1)
    centers : RBF centres,  shape (K, 1)

    Returns
    -------
    Phi : design matrix,    shape (N, K)
    """
    # Broadcasting: X has shape (N,1), centers.T has shape (1,K)
    squared_distances = (X - centers.T) ** 2   # (N, K)
    Phi = np.exp(-gamma * squared_distances)
    return Phi


def fit(Phi: np.ndarray, y: np.ndarray) -> np.ndarray:
    """
    Solve for output weights using the pseudoinverse (least-squares).
    w = Phi^+ @ y
    """
    w = np.linalg.pinv(Phi) @ y
    return w


def predict(Phi: np.ndarray, w: np.ndarray) -> np.ndarray:
    return Phi @ w


if __name__ == "__main__":
    centers = make_centers()
    Phi     = rbf_layer(X, centers)

    print(f"Input shape  : {X.shape}")
    print(f"Phi shape    : {Phi.shape}  (samples x centres)")

    w    = fit(Phi, y)
    pred = predict(Phi, w)

    mse = float(np.mean((pred - y) ** 2))
    print(f"MSE on training data    : {mse:.6f}")
    print(f"First 5 predictions     : {np.round(pred[:5], 3)}")
    print(f"First 5 true values     : {np.round(y[:5], 3)}")

```
