```python
# Output Guide:
# - Prints SOM training message and corner neuron weights.
# - Correctness check: corner weights should spread toward distinct data clusters.
# - This signifies topology-preserving organization of the map during unsupervised learning.
"""
Self-Organizing Map (SOM) — NumPy From Scratch
===============================================
A SOM is an unsupervised neural network that learns a low-dimensional
(here: 2-D grid) topology-preserving map of the input space.

Training algorithm (competitive learning):
    1. Pick a random input sample x.
    2. Find the Best Matching Unit (BMU): the neuron whose weight vector
       is closest to x (Euclidean distance).
    3. Update the BMU and its grid neighbours toward x:
           W[i] += lr * h(i, bmu) * (x - W[i])
       where h is a Gaussian neighbourhood function that decays over time.

Dataset: two 2-D Gaussian clusters at (-1,-1) and (1,1).
Grid:    5×5 neurons.
"""

import numpy as np

rng = np.random.default_rng(42)

# ── Data ─────────────────────────────────────────────────────────────────────
cluster_a = rng.normal([-1, -1], 0.25, size=(60, 2))
cluster_b = rng.normal([ 1,  1], 0.25, size=(60, 2))
X = np.vstack([cluster_a, cluster_b])   # (120, 2)

# ── SOM grid ──────────────────────────────────────────────────────────────────
GRID_SIZE  = 5
N_EPOCHS   = 500
INIT_LR    = 0.3
INIT_WIDTH = 2.0   # initial neighbourhood radius

# Initialise weight vectors randomly
W = rng.normal(size=(GRID_SIZE, GRID_SIZE, 2))

# Pre-compute (i,j) grid coordinates for each neuron
grid_i, grid_j = np.meshgrid(np.arange(GRID_SIZE), np.arange(GRID_SIZE), indexing='ij')
grid_coords = np.stack([grid_i, grid_j], axis=-1)   # (5, 5, 2)


def find_bmu(x: np.ndarray, W: np.ndarray):
    """
    Find the Best Matching Unit index for input x.

    Returns the (row, col) index of the neuron closest to x.
    """
    distances = ((W - x) ** 2).sum(axis=-1)   # (GRID_SIZE, GRID_SIZE)
    bmu_flat  = np.argmin(distances)
    bmu       = np.array(np.unravel_index(bmu_flat, (GRID_SIZE, GRID_SIZE)))
    return bmu


def neighbourhood(bmu, grid_coords, width: float) -> np.ndarray:
    """
    Gaussian neighbourhood influence centred on the BMU.

    Returns an array of shape (GRID_SIZE, GRID_SIZE) with values in (0, 1].
    """
    sq_dist = ((grid_coords - bmu) ** 2).sum(axis=-1)   # squared grid distance
    return np.exp(-sq_dist / (2 * width ** 2))


def train_som(X, W, n_epochs=N_EPOCHS):
    """Run the SOM training loop."""
    for t in range(n_epochs):
        # ── Anneal learning rate and neighbourhood width linearly ──────────
        lr    = INIT_LR    * (1 - t / n_epochs)
        width = max(0.4, INIT_WIDTH * (1 - t / n_epochs))

        # ── Pick a random training sample ─────────────────────────────────
        x = X[rng.integers(len(X))]

        # ── Find BMU ──────────────────────────────────────────────────────
        bmu = find_bmu(x, W)

        # ── Compute neighbourhood influence ───────────────────────────────
        h = neighbourhood(bmu, grid_coords, width)   # (GRID_SIZE, GRID_SIZE)

        # ── Update all neurons (strong near BMU, weak far away) ───────────
        W += lr * h[..., None] * (x - W)

    return W


if __name__ == "__main__":
    print(f"Training SOM ({GRID_SIZE}×{GRID_SIZE} grid) on {len(X)} samples …")
    W = train_som(X, W)

    print(f"Corner neuron weights (should cluster near ±1):")
    corners = W[[0, -1]][:, [0, -1]]   # (2, 2, 2)
    print(np.round(corners, 3))

```
