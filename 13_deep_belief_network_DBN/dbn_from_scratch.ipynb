```python
# Output Guide:
# - Prints RBM training message, feature matrix shape, and first feature vector.
# - Correctness check: feature shape should be (num_samples, hidden_dim).
# - Output features signify learned higher-level representations from the RBM layer.
"""
Deep Belief Network (DBN) — NumPy From Scratch
================================================
A DBN is a stack of Restricted Boltzmann Machines (RBMs) trained
greedily, one layer at a time.

This file demonstrates a single-layer RBM trained with
Contrastive Divergence 1 (CD-1):

    1. Positive phase  — clamp visible v, sample hidden h0 ~ p(h|v)
    2. Negative phase  — reconstruct visible v1 ~ p(v|h0),
                         then sample hidden h1 ~ p(h|v1)
    3. Update weights  — W += lr * (v^T h0 - v1^T h1) / N

This is the building block for stacking multiple RBMs into a DBN.
"""

import numpy as np

rng = np.random.default_rng(42)

# ── Data ─────────────────────────────────────────────────────────────────────
N_SAMPLES   = 100
VISIBLE_DIM = 6
HIDDEN_DIM  = 4
EPOCHS      = 300
LR          = 0.05

# Binary visible data (e.g., binary pixel patterns)
V = (rng.random((N_SAMPLES, VISIBLE_DIM)) > 0.5).astype(float)


def sigmoid(x: np.ndarray) -> np.ndarray:
    return 1.0 / (1.0 + np.exp(-x))


def init_rbm():
    """Initialise RBM weights and biases."""
    W  = rng.normal(scale=0.1, size=(VISIBLE_DIM, HIDDEN_DIM))  # weight matrix
    v_bias = np.zeros(VISIBLE_DIM)    # visible bias
    h_bias = np.zeros(HIDDEN_DIM)     # hidden bias
    return W, v_bias, h_bias


def sample_hidden(V, W, h_bias):
    """
    Positive phase: compute p(h=1|v) and draw a binary sample.

    Returns
    -------
    p_h : probabilities, shape (N, HIDDEN_DIM)
    h   : binary samples, shape (N, HIDDEN_DIM)
    """
    p_h = sigmoid(V @ W + h_bias)
    h   = (rng.random(p_h.shape) < p_h).astype(float)
    return p_h, h


def sample_visible(H, W, v_bias):
    """
    Negative phase step 1: reconstruct visible from hidden.

    Returns
    -------
    p_v : probabilities, shape (N, VISIBLE_DIM)
    v1  : binary samples, shape (N, VISIBLE_DIM)
    """
    p_v = sigmoid(H @ W.T + v_bias)
    v1  = (rng.random(p_v.shape) < p_v).astype(float)
    return p_v, v1


def train_rbm(V, W, v_bias, h_bias, epochs=EPOCHS, lr=LR):
    """
    Train the RBM using Contrastive Divergence (CD-1).
    """
    n = len(V)
    for epoch in range(epochs):
        # ── Positive phase ────────────────────────────────────────────────
        p_h0, h0 = sample_hidden(V, W, h_bias)     # hidden given real data

        # ── Negative phase ────────────────────────────────────────────────
        _, v1     = sample_visible(h0, W, v_bias)   # reconstruct visible
        p_h1, _   = sample_hidden(v1, W, h_bias)    # hidden given reconstruction

        # ── CD-1 weight update ────────────────────────────────────────────
        # Positive associations minus negative associations
        W      += lr * (V.T @ p_h0 - v1.T @ p_h1) / n
        v_bias += lr * np.mean(V - v1, axis=0)
        h_bias += lr * np.mean(p_h0 - p_h1, axis=0)

    return W, v_bias, h_bias


if __name__ == "__main__":
    W, v_bias, h_bias = init_rbm()

    print(f"Training RBM (visible={VISIBLE_DIM}, hidden={HIDDEN_DIM}) …")
    W, v_bias, h_bias = train_rbm(V, W, v_bias, h_bias)

    # Extract learned features (hidden activations on training data)
    features, _ = sample_hidden(V, W, h_bias)
    print(f"Feature shape       : {features.shape}  (N x hidden_dim)")
    print(f"First sample features: {np.round(features[0], 3)}")

```
