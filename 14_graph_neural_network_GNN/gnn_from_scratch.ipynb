```python
# Output Guide:
# - Prints node features, normalized adjacency, and node embeddings after one GCN layer.
# - Correctness check: normalized adjacency should be symmetric for undirected graphs.
# - Embeddings signify neighborhood-aware feature aggregation per node.
"""
Graph Neural Network (GNN) — NumPy From Scratch
================================================
Implements one layer of a Graph Convolutional Network (GCN).

GCN layer equation:
    H = ReLU( D̂^{-1/2} A_hat D̂^{-1/2} X W )

Where:
    A_hat  = A + I        (adjacency with added self-loops)
    D̂  = degree matrix of A_hat
    X  = node feature matrix   (N, in_features)
    W  = learnable weight matrix (in_features, out_features)

The symmetric normalisation D̂^{-1/2} A_hat D̂^{-1/2} prevents
feature explosion in nodes with many neighbours.

Graph: 4-node fully-connected example (minus some edges).
"""

import numpy as np

rng = np.random.default_rng(42)

N_NODES     = 4
IN_FEATURES = 3
OUT_FEATURES = 2


def build_graph():
    """
    Build a simple 4-node graph adjacency matrix.
    Nodes 0-1-2-3 with some edges removed.
    """
    A = np.array([
        [1, 1, 1, 0],
        [1, 1, 0, 1],
        [1, 0, 1, 1],
        [0, 1, 1, 1],
    ], dtype=float)
    return A


def normalise_adjacency(A: np.ndarray) -> np.ndarray:
    """
    Compute the symmetrically normalised adjacency: D̂^{-1/2} A D̂{-1/2}.

    Note: self-loops are already included in A (diagonal = 1).
    """
    degree = A.sum(axis=1)                      # degree of each node
    D_inv_sqrt = np.diag(1.0 / np.sqrt(degree)) # D^{-1/2}
    A_norm = D_inv_sqrt @ A @ D_inv_sqrt         # symmetric normalisation
    return A_norm


def gcn_layer(A_norm: np.ndarray, X: np.ndarray, W: np.ndarray) -> np.ndarray:
    """
    One GCN message-passing layer.

    Parameters
    ----------
    A_norm : normalised adjacency, shape (N, N)
    X      : node features,        shape (N, in_features)
    W      : weight matrix,        shape (in_features, out_features)

    Returns
    -------
    H : new node embeddings,       shape (N, out_features)
    """
    # Aggregate neighbour features, then linearly transform
    aggregated = A_norm @ X     # neighbourhood aggregation: (N, in_features)
    H = np.maximum(0, aggregated @ W)   # linear transform + ReLU
    return H


if __name__ == "__main__":
    # Graph and features
    A = build_graph()
    X = rng.normal(size=(N_NODES, IN_FEATURES))    # random node features
    W = rng.normal(scale=0.4, size=(IN_FEATURES, OUT_FEATURES))

    print(f"Node features X :\n{np.round(X, 3)}\n")

    # Normalise adjacency
    A_norm = normalise_adjacency(A)
    print(f"Normalised adjacency:\n{np.round(A_norm, 3)}\n")

    # One GCN forward pass
    H = gcn_layer(A_norm, X, W)
    print(f"Node embeddings H (after 1 GCN layer):\n{np.round(H, 3)}")

```
