```python
# Output Guide:
# - Prints spike train summary plus a per-timestep voltage/spike table.
# - Correctness check: spikes occur when membrane voltage reaches threshold and then resets.
# - Total spikes/firing rate quantify neuron response strength to input rates.
"""
Spiking Neural Network (SNN) — NumPy From Scratch
===================================================
Implements the Leaky Integrate-and-Fire (LIF) neuron model.

The LIF model is the simplest biologically plausible neuron:

    Membrane update:
        V[t] = tau * V[t-1] + I[t]    (leak + integrate input current)

    Spike condition:
        if V[t] >= V_threshold:
            spike = 1
            V[t]  = 0                  (reset to resting potential)

    Where tau < 1 is the leak factor — the neuron "forgets" old voltage.

Here we simulate a single LIF neuron receiving Poisson-like input
with linearly increasing firing rates over 20 time steps.
"""

import numpy as np

rng = np.random.default_rng(42)

# ── LIF neuron parameters ─────────────────────────────────────────────────────
TAU         = 0.85    # membrane time constant (leak factor)
V_THRESHOLD = 1.0     # spike threshold
V_RESET     = 0.0     # voltage after a spike

# ── Input current ──────────────────────────────────────────────────────────────
# Poisson-like: input is 1 with probability equal to the firing rate
firing_rates = np.linspace(0.05, 0.9, 20)   # increasing rates over time


def lif_neuron(firing_rates, tau=TAU, v_threshold=V_THRESHOLD, v_reset=V_RESET):
    """
    Simulate a single LIF neuron over a sequence of time steps.

    Parameters
    ----------
    firing_rates : array of input Poisson rates, one per time step
    tau          : membrane leak factor (0 < tau < 1)
    v_threshold  : voltage threshold to trigger a spike
    v_reset      : voltage to reset to after spiking

    Returns
    -------
    spike_train : list of 0/1 values, one per time step
    voltages    : membrane voltage before each spike decision
    """
    v          = 0.0    # initial membrane voltage
    spike_train = []
    voltages    = []

    for rate in firing_rates:
        # Leaky integration: decay existing voltage + add binary input current
        input_current = float(rng.random() < rate)   # Poisson spike with given rate
        v = tau * v + input_current

        # Record voltage before threshold check
        voltages.append(round(v, 4))

        # Threshold check: fire if voltage crosses threshold
        if v >= v_threshold:
            spike_train.append(1)
            v = v_reset    # reset membrane potential
        else:
            spike_train.append(0)

    return spike_train, voltages


if __name__ == "__main__":
    spike_train, voltages = lif_neuron(firing_rates)

    print("LIF Neuron Simulation")
    print(f"Time steps    : {len(firing_rates)}")
    print(f"Spike train   : {spike_train}")
    print(f"Total spikes  : {sum(spike_train)}")
    print(f"Firing rate   : {sum(spike_train) / len(spike_train):.2f} spikes/step")
    print(f"\nTime | Rate  | Voltage | Spike")
    print("-" * 36)
    for t, (rate, v, s) in enumerate(zip(firing_rates, voltages, spike_train)):
        spike_marker = " <<< SPIKE" if s else ""
        print(f"  {t:2d} | {rate:.2f}  | {v:+.4f}  | {s}{spike_marker}")

```
