```python
# Output Guide:
# - Prints input/output shapes, L2 norms, and sample rows.
# - Correctness check: output shape should match input shape due to skip connection.
# - Differences between sample input/output indicate residual transformation effects.
"""
Residual Network (ResNet) — NumPy From Scratch
===============================================
Demonstrates the key idea of ResNet: the identity skip connection.

A standard residual block computes:
    H(x) = F(x) + x          ← skip connection adds the input directly

Where F(x) is the "residual mapping":
    F(x) = ReLU( x @ W1 ) @ W2    (two linear layers, one ReLU in between)

And the final output uses one more ReLU:
    output = ReLU( F(x) + x )

Why skip connections help:
  - If F(x) ≈ 0, the block becomes the identity — no degradation.
  - Gradients bypass deep layers → mitigates vanishing gradient.
  - The network only needs to learn the *residual* adjustment.
"""

import numpy as np

rng = np.random.default_rng(42)

BATCH_SIZE = 3
DIM        = 6    # input/output dimension (must match for identity skip)


def relu(z: np.ndarray) -> np.ndarray:
    """ReLU activation: max(0, z)."""
    return np.maximum(0, z)


def init_residual_block(dim: int = DIM):
    """Initialise two weight matrices for the residual mapping F(x)."""
    scale = 0.2
    W1 = rng.normal(scale=scale, size=(dim, dim))
    W2 = rng.normal(scale=scale, size=(dim, dim))
    return W1, W2


def residual_block(X: np.ndarray, W1: np.ndarray, W2: np.ndarray) -> np.ndarray:
    """
    Forward pass through one residual block.

    F(x) = ReLU(x @ W1) @ W2        ← residual mapping
    H(x) = ReLU( F(x) + x )         ← skip connection + activation

    Parameters
    ----------
    X  : input batch, shape (batch, dim)
    W1 : first layer weights,  shape (dim, dim)
    W2 : second layer weights, shape (dim, dim)

    Returns
    -------
    output : shape (batch, dim) — same as input (skip connection preserved shape)
    """
    # First linear layer + ReLU
    F = relu(X @ W1)        # (batch, dim)

    # Second linear layer (no activation yet)
    F = F @ W2              # (batch, dim)

    # Skip connection: add input directly to the residual output
    residual_sum = F + X    # (batch, dim)  ← the key ResNet operation

    # Final ReLU after the skip addition
    output = relu(residual_sum)

    return output


if __name__ == "__main__":
    X = rng.normal(size=(BATCH_SIZE, DIM))   # random input batch
    W1, W2 = init_residual_block()

    output = residual_block(X, W1, W2)

    print(f"Input  shape : {X.shape}")
    print(f"Output shape : {output.shape}  (unchanged — skip connection)")
    print(f"\nInput  L2 norm : {np.linalg.norm(X):.4f}")
    print(f"Output L2 norm : {np.linalg.norm(output):.4f}")
    print(f"\nSample input  row 0: {np.round(X[0], 3)}")
    print(f"Sample output row 0: {np.round(output[0], 3)}")

```
