```python
# Output Guide:
# - Prints clean signal, beta schedule, and selected forward-diffusion timesteps.
# - Correctness check: signal coefficient decreases while noise coefficient increases with t.
# - At large t, x_t should resemble Gaussian noise, showing progressive corruption.
"""
Denoising Diffusion Probabilistic Model (DDPM) — NumPy From Scratch
====================================================================
Implements the DDPM *forward* (noising) process.

The forward process gradually adds Gaussian noise to a clean sample x0
over T = 100 timesteps according to a noise schedule β_1, …, β_T.

At each step t, the noisy sample is:
    x_t = sqrt(ᾱ_t) * x0 + sqrt(1 - ᾱ_t) * ε,   ε ~ N(0, I)

Where:
    α_t  = 1 - β_t          (signal retention at step t)
    ᾱ_t  = ∏_{s=1}^{t} α_s  (cumulative product — total signal retention)

As t → T:
    sqrt(ᾱ_T) → 0   → x_T ≈ pure noise
    sqrt(1 - ᾱ_T) → 1

The reverse process (denoising, not implemented here) learns to undo
this noising one step at a time.
"""

import numpy as np

rng = np.random.default_rng(42)

# ── Noise schedule ────────────────────────────────────────────────────────────
T     = 100                                    # total diffusion timesteps
betas = np.linspace(1e-4, 0.02, T)            # linear β schedule: small → larger
alphas = 1.0 - betas                           # α_t = 1 - β_t
alpha_bar = np.cumprod(alphas)                 # ᾱ_t = cumulative product of α

# ── Clean signal x0 ───────────────────────────────────────────────────────────
x0 = np.linspace(-1, 1, 8)                    # 8-element clean "image"


def q_sample(x0: np.ndarray, t: int, alpha_bar: np.ndarray) -> np.ndarray:
    """
    Sample x_t from q(x_t | x0) using the closed-form forward process.

    x_t = sqrt(ᾱ_t) * x0 + sqrt(1 - ᾱ_t) * ε

    Parameters
    ----------
    x0        : clean sample
    t         : timestep index (0-based)
    alpha_bar : cumulative product of alphas, shape (T,)

    Returns
    -------
    x_t     : noisy sample at timestep t
    epsilon : the noise that was added (useful for training the denoiser)
    """
    signal_scale = np.sqrt(alpha_bar[t])          # how much of x0 remains
    noise_scale  = np.sqrt(1.0 - alpha_bar[t])    # how much noise is mixed in

    epsilon = rng.normal(size=x0.shape)            # ε ~ N(0, I)
    x_t     = signal_scale * x0 + noise_scale * epsilon

    return x_t, epsilon


if __name__ == "__main__":
    print(f"Clean signal x0 : {np.round(x0, 3)}")
    print(f"Noise schedule  : beta from {betas[0]:.4f} to {betas[-1]:.4f}")
    print(f"\nForward diffusion at selected timesteps:")
    print(f"{'t':>4} | {'sqrt(a_bar_t) signal':>20} | {'sqrt(1-a_bar_t) noise':>21} | x_t")
    print("-" * 75)

    for t in [0, 25, 50, 74, 99]:
        x_t, _ = q_sample(x0, t, alpha_bar)
        sig  = np.sqrt(alpha_bar[t])
        noise = np.sqrt(1.0 - alpha_bar[t])
        print(f"{t:>4} | {sig:>14.4f} | {noise:>17.4f} | {np.round(x_t, 3)}")

    print(f"\nAt t=99: signal ~= 0, x_t ~= pure Gaussian noise.")

```
